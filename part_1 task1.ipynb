```python
# Task 1: Variables and types, one order with six items

order_id = 101
is_cancelled = False

items = [
    {"product": "Wireless Mouse", "quantity": 2,  "unit_price": 799.50},
    {"product": "USB Keyboard",   "quantity": 1,  "unit_price": 1250.00},
    {"product": "Notebook Pack",  "quantity": 12, "unit_price": 45.75},
    {"product": "HDMI Cable",     "quantity": 3,  "unit_price": 299.99},
    {"product": "Laptop Stand",   "quantity": 1,  "unit_price": 1899.00},
    {"product": "Webcam",         "quantity": 4,  "unit_price": 2499.50},
]

print(f"order_id: {order_id} | Type: {type(order_id)}")
print(f"is_cancelled: {is_cancelled} | Type: {type(is_cancelled)}")
print()

for number, item in enumerate(items, start=1):
    print(f"--- Item {number} ---")
    print(f"product: {item['product']} | Type: {type(item['product'])}")
    print(f"quantity: {item['quantity']} | Type: {type(item['quantity'])}")
    print(f"unit_price: {item['unit_price']} | Type: {type(item['unit_price'])}")
    print()
```
