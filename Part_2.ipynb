```python

#Task 1.
# billing.py

def calculate_subtotal(quantity, unit_price):
    return quantity * unit_price


def apply_discount(subtotal):
    if subtotal >= 10000:
        discount_rate = 0.10
    elif subtotal >= 5000:
        discount_rate = 0.05
    else:
        discount_rate = 0.00
    return subtotal * discount_rate


def calculate_gst(amount, rate=0.18):
    return amount * rate


# Call the functions to see output
sub = calculate_subtotal(8, 650)
disc = apply_discount(sub)
gst = calculate_gst(sub - disc)
total = (sub - disc) + gst

print("Subtotal:", sub)
print("Discount:", disc)
print("GST:", gst)
print("Final Total:", round(total, 2))
```

```python
# Task 2.
def generate_invoice(customer, *items, **options):
    gst_rate = options.get("gst_rate", 0.18)
    currency = options.get("currency", "Rs.")

    invoice_lines = []
    grand_total = 0.0

    for product, qty, price in items:
        if qty <= 0:
            continue

        subtotal = qty * price

        if subtotal >= 10000:
            discount = subtotal * 0.10
        elif subtotal >= 5000:
            discount = subtotal * 0.05
        else:
            discount = 0.0

        taxable = subtotal - discount
        gst = taxable * gst_rate
        line_total = round(taxable + gst, 2)
        grand_total += line_total

        invoice_lines.append({
            "product": product,
            "qty": qty,
            "line_total": f"{currency} {line_total:.2f}"
        })

    return {
        "customer": customer,
        "lines": invoice_lines,
        "grand_total": f"{currency} {round(grand_total, 2):.2f}"
    }


result = generate_invoice("Aarav", ("Wireless Mouse", 2, 799), ("Basmati Rice 5kg", 8, 650))
print(result)
```

```python
# Task 3.
import math
from datetime import datetime

# Sample invoice lines (product, line_total)
invoice_lines = [
    {"product": "Wireless Mouse", "line_total": 1885.64},
    {"product": "Basmati Rice 5kg", "line_total": 5829.20},
    {"product": "Denim Jeans", "line_total": 11886.97},
    {"product": "Notebook Pack", "line_total": 1416.00}
]

# 1. Sort invoice lines by line_total in descending order using lambda
sorted_lines = sorted(invoice_lines, key=lambda item: item["line_total"], reverse=True)

# 2. Calculate raw total and round up to the next rupee using math.ceil()
raw_total = sum(item["line_total"] for item in sorted_lines)
payable_amount = math.ceil(raw_total)

# 3. Get and format the current invoice date using datetime
invoice_date = datetime.now().strftime("%Y-%m-%d")

# Print results
print("Invoice Date:", invoice_date)
print("\nSorted Invoice Lines (Highest to Lowest):")
for line in sorted_lines:
    print(f"- {line['product']}: Rs. {line['line_total']:.2f}")

print(f"\nRaw Total: Rs. {raw_total:.2f}")
print(f"Payable Amount (Rounded Up): Rs. {payable_amount}")
```

```python
#Task 4.
# Billing functions from earlier task
def calculate_subtotal(quantity, unit_price):
    return quantity * unit_price


def apply_discount(subtotal):
    if subtotal >= 10000:
        discount_rate = 0.10
    elif subtotal >= 5000:
        discount_rate = 0.05
    else:
        discount_rate = 0.00
    return subtotal * discount_rate


def calculate_gst(amount, rate=0.18):
    return amount * rate


# 1. Product Class
class Product:
    def __init__(self, name, category, unit_price):
        self.name = name
        self.category = category
        self.unit_price = unit_price

    def __str__(self):
        return f"{self.name} ({self.category}) - Rs. {self.unit_price:.2f}"


# 2. Order Class
class Order:
    def __init__(self, order_id, product, quantity):
        self.order_id = order_id
        self.product = product
        self.quantity = quantity

    def total(self):
        subtotal = calculate_subtotal(self.quantity, self.product.unit_price)
        discount = apply_discount(subtotal)
        taxable = subtotal - discount
        gst = calculate_gst(taxable)
        return round(taxable + gst, 2)

    @classmethod
    def from_row(cls, row):
        product = Product(
            name=row["name"],
            category=row["category"],
            unit_price=float(row["unit_price"])
        )
        return cls(
            order_id=row["order_id"],
            product=product,
            quantity=int(row["quantity"])
        )


# Test the classes and print output
p1 = Product("Wireless Mouse", "Electronics", 799)
o1 = Order("ORD-101", p1, 2)

print("Product:", p1)
print(f"Order {o1.order_id} Total: Rs. {o1.total():.2f}")

# Test creating an Order from a dictionary row
sample_row = {
    "order_id": "ORD-102",
    "name": "Basmati Rice 5kg",
    "category": "Groceries",
    "unit_price": 650,
    "quantity": 8
}

o2 = Order.from_row(sample_row)
print("\nProduct from row:", o2.product)
print(f"Order {o2.order_id} Total: Rs. {o2.total():.2f}")
```

```python
#Task 5.
# Billing functions
def calculate_subtotal(quantity, unit_price):
    return quantity * unit_price


def apply_discount(subtotal):
    if subtotal >= 10000:
        discount_rate = 0.10
    elif subtotal >= 5000:
        discount_rate = 0.05
    else:
        discount_rate = 0.00
    return subtotal * discount_rate


def calculate_gst(amount, rate=0.18):
    return amount * rate


# Base Product Class
class Product:
    def __init__(self, name, category, unit_price):
        self.name = name
        self.category = category
        self.unit_price = unit_price

    def __str__(self):
        return f"{self.name} ({self.category}) - Rs. {self.unit_price:.2f}"


# Parent Order Class
class Order:
    def __init__(self, order_id, product, quantity):
        self.order_id = order_id
        self.product = product
        self.quantity = quantity

    def total(self):
        subtotal = calculate_subtotal(self.quantity, self.product.unit_price)
        discount = apply_discount(subtotal)
        taxable = subtotal - discount
        gst = calculate_gst(taxable)
        return round(taxable + gst, 2)


# Child PremiumOrder Class (Inheritance)
class PremiumOrder(Order):
    def total(self):
        # Call parent total() using super()
        standard_total = super().total()

        # Since standard_total already includes 18% GST, reducing the pre-GST
        # taxable amount by 3% scales the final total by 0.97
        premium_total = standard_total * 0.97
        return round(premium_total, 2)


# Create products and a mixed list of Order and PremiumOrder objects
p1 = Product("Wireless Mouse", "Electronics", 799)
p2 = Product("Basmati Rice 5kg", "Groceries", 650)
p3 = Product("Denim Jeans", "Apparel", 1599)

orders_list = [
    Order("ORD-101", p1, 2),          # Standard Order
    PremiumOrder("ORD-102", p2, 8),   # Premium Order (extra 3% off)
    Order("ORD-103", p3, 7),          # Standard Order
    PremiumOrder("ORD-104", p1, 2)    # Premium Order (extra 3% off)
]

# Demonstrate polymorphism by calling .total() on each item in the mixed list
for order in orders_list:
    order_type = type(order).__name__
    print(f"{order.order_id} [{order_type}] - {order.product.name}: Rs. {order.total():.2f}")
```

```python
# TASK 6.
class Product:
    def __init__(self, name, stock):
        self.name = name
        # Private attribute using double underscore
        self.__stock = stock

    # Exposing the private __stock attribute via @property
    @property
    def stock(self):
        return self.__stock

    # Method to reduce stock and raise ValueError if insufficient
    def reduce_stock(self, qty):
        if qty <= 0:
            raise ValueError("Quantity to reduce must be greater than zero.")
        if qty > self.__stock:
            raise ValueError(f"Insufficient stock! Available: {self.__stock}, Requested: {qty}")
        
        self.__stock -= qty
        print(f"Successfully reduced stock by {qty}. Remaining stock: {self.__stock}")


# --- Demonstration using try/except ---
item = Product("Laptop", 10)
print(f"Product: {item.name} | Initial Stock: {item.stock}\n")

# Case 1: Valid stock reduction
try:
    item.reduce_stock(4)
except ValueError as error:
    print(f"Error: {error}")

# Case 2: Insufficient stock (triggers ValueError)
try:
    item.reduce_stock(8)
except ValueError as error:
    print(f"Error: {error}")

print(f"\nFinal Stock: {item.stock}")
```
