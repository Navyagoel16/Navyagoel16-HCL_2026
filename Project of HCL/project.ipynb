```python
from datetime import datetime
import numpy as np
import pandas as pd

# ==============================================================================
# SETUP: Built-in Sample Dataset (No external CSV needed)
# ==============================================================================
raw_data = [
    {
        'Order_ID': 'ORD001',
        'Order_Date': '2025-01-15',
        'Customer_ID': 'CUST01',
        'City': 'Mumbai',
        'Product': 'Laptop',
        'Category': 'Electronics',
        'Quantity': 1,
        'Unit_Price': 60000,
        'Revenue': 60000,
        'Payment_Method': 'Credit Card',
    },
    {
        'Order_ID': 'ORD002',
        'Order_Date': '2025-02-10',
        'Customer_ID': 'CUST02',
        'City': 'Delhi',
        'Product': 'Headphones',
        'Category': 'Electronics',
        'Quantity': 2,
        'Unit_Price': 2000,
        'Revenue': 4000,
        'Payment_Method': 'UPI',
    },
    {
        'Order_ID': 'ORD003',
        'Order_Date': '2025-03-05',
        'Customer_ID': 'CUST03',
        'City': None,
        'Product': 'Office Chair',
        'Category': 'Furniture',
        'Quantity': 1,
        'Unit_Price': 8500,
        'Revenue': 8500,
        'Payment_Method': 'Debit Card',
    },
    {
        'Order_ID': 'ORD004',
        'Order_Date': '2025-04-18',
        'Customer_ID': 'CUST01',
        'City': 'Mumbai',
        'Product': 'Mouse',
        'Category': 'Accessories',
        'Quantity': 3,
        'Unit_Price': 800,
        'Revenue': 2400,
        'Payment_Method': 'UPI',
    },
    {
        'Order_ID': 'ORD005',
        'Order_Date': '2025-05-22',
        'Customer_ID': 'CUST04',
        'City': 'Bengaluru',
        'Product': 'Monitor',
        'Category': 'Electronics',
        'Quantity': 2,
        'Unit_Price': 15000,
        'Revenue': 30000,
        'Payment_Method': 'Credit Card',
    },
    {
        'Order_ID': 'ORD006',
        'Order_Date': '2025-06-11',
        'Customer_ID': 'CUST05',
        'City': 'Mumbai',
        'Product': 'Keyboard',
        'Category': 'Accessories',
        'Quantity': 2,
        'Unit_Price': 1500,
        'Revenue': 3000,
        'Payment_Method': 'UPI',
    },
    {
        'Order_ID': 'ORD007',
        'Order_Date': '2025-07-19',
        'Customer_ID': 'CUST06',
        'City': 'Pune',
        'Product': 'Desk Lamp',
        'Category': 'Furniture',
        'Quantity': 4,
        'Unit_Price': 1200,
        'Revenue': 4800,
        'Payment_Method': 'Cash',
    },
    {
        'Order_ID': 'ORD008',
        'Order_Date': '2025-08-25',
        'Customer_ID': 'CUST07',
        'City': 'Delhi',
        'Product': 'Smartphone',
        'Category': 'Electronics',
        'Quantity': 1,
        'Unit_Price': 25000,
        'Revenue': 25000,
        'Payment_Method': 'Credit Card',
    },
    {
        'Order_ID': 'ORD009',
        'Order_Date': '2025-09-14',
        'Customer_ID': 'CUST08',
        'City': None,
        'Product': 'Backpack',
        'Category': 'Accessories',
        'Quantity': 2,
        'Unit_Price': 1800,
        'Revenue': 3600,
        'Payment_Method': 'UPI',
    },
    {
        'Order_ID': 'ORD010',
        'Order_Date': '2025-10-30',
        'Customer_ID': 'CUST09',
        'City': 'Bengaluru',
        'Product': 'Tablet',
        'Category': 'Electronics',
        'Quantity': 1,
        'Unit_Price': 22000,
        'Revenue': 22000,
        'Payment_Method': 'Debit Card',
    },
    {
        'Order_ID': 'ORD011',
        'Order_Date': '2025-11-12',
        'Customer_ID': 'CUST10',
        'City': 'Mumbai',
        'Product': 'Smartwatch',
        'Category': 'Electronics',
        'Quantity': 3,
        'Unit_Price': 5000,
        'Revenue': 15000,
        'Payment_Method': 'UPI',
    },
    {
        'Order_ID': 'ORD012',
        'Order_Date': '2025-11-25',
        'Customer_ID': 'CUST11',
        'City': 'Delhi',
        'Product': 'Gaming Chair',
        'Category': 'Furniture',
        'Quantity': 5,
        'Unit_Price': 18000,
        'Revenue': 90000,
        'Payment_Method': 'Credit Card',
    },  # Large outlier order
    {
        'Order_ID': 'ORD013',
        'Order_Date': '2025-12-08',
        'Customer_ID': 'CUST12',
        'City': 'Pune',
        'Product': 'USB Hub',
        'Category': 'Accessories',
        'Quantity': 3,
        'Unit_Price': 900,
        'Revenue': 2700,
        'Payment_Method': 'UPI',
    },
    # Duplicate order added intentionally to demonstrate Task 5
    {
        'Order_ID': 'ORD002',
        'Order_Date': '2025-02-10',
        'Customer_ID': 'CUST02',
        'City': 'Delhi',
        'Product': 'Headphones',
        'Category': 'Electronics',
        'Quantity': 2,
        'Unit_Price': 2000,
        'Revenue': 4000,
        'Payment_Method': 'UPI',
    },
]

# ==============================================================================
# TASK 4: Load data and inspect dimensions, columns, and data types
# ==============================================================================
df = pd.DataFrame(raw_data)
print('=== TASK 4: DATASET OVERVIEW ===')
print('Dimensions (Rows, Columns):', df.shape)
print('\nColumns and Data Types:')
print(df.dtypes)

# ==============================================================================
# TASK 5: Identify missing values and duplicate orders
# ==============================================================================
print('\n=== TASK 5: MISSING VALUES & DUPLICATES ===')
print('Missing values per column:')
print(df.isnull().sum())

duplicate_count = df.duplicated(subset=['Order_ID']).sum()
print(f'\nDuplicate orders found: {duplicate_count}')
df = df.drop_duplicates(subset=['Order_ID']).copy()

# ==============================================================================
# TASK 6: Clean missing City values using a justified strategy (Mode)
# ==============================================================================
# Justification: City is categorical data. Replacing missing values with the mode
# (most frequent city) retains all order records and revenue without skewing totals.
mode_city = df['City'].mode()[0]
df['City'] = df['City'].fillna(mode_city)
print(
    f"\n=== TASK 6: MISSING CITY CLEANED (Replaced with Mode: '{mode_city}') ==="
)

# ==============================================================================
# TASK 7: Convert Order_Date to datetime and extract month and weekday
# ==============================================================================
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df['Month'] = df['Order_Date'].dt.month
df['Month_Name'] = df['Order_Date'].dt.month_name()
df['Weekday'] = df['Order_Date'].dt.day_name()
print('\n=== TASK 7: EXTRACTED MONTH & WEEKDAY (First 5 Rows) ===')
print(df[['Order_ID', 'Order_Date', 'Month_Name', 'Weekday']].head())

# ==============================================================================
# TASK 8: Calculate row-wise revenue and verify the supplied Revenue column
# ==============================================================================
df['Calculated_Revenue'] = df['Quantity'] * df['Unit_Price']
mismatches = (df['Calculated_Revenue'] != df['Revenue']).sum()
print('\n=== TASK 8: REVENUE VERIFICATION ===')
print(f'Mismatched rows between Calculated and Supplied Revenue: {mismatches}')

# ==============================================================================
# TASK 9: Find total revenue, total quantity, and average order value
# ==============================================================================
total_revenue = df['Revenue'].sum()
total_quantity = df['Quantity'].sum()
avg_order_value = df['Revenue'].mean()
print('\n=== TASK 9: OVERALL METRICS ===')
print(f'Total Revenue:       ₹{total_revenue:,.2f}')
print(f'Total Quantity Sold: {total_quantity}')
print(f'Average Order Value: ₹{avg_order_value:,.2f}')

# ==============================================================================
# TASK 10: Find the 10 products with the highest revenue
# ==============================================================================
top_10_products = (
    df.groupby('Product')['Revenue'].sum().sort_values(ascending=False).head(10)
)
print('\n=== TASK 10: TOP 10 PRODUCTS BY REVENUE ===')
print(top_10_products)

# ==============================================================================
# TASK 11: Rank cities by revenue
# ==============================================================================
city_ranking = df.groupby('City')['Revenue'].sum().sort_values(ascending=False)
print('\n=== TASK 11: CITIES RANKED BY REVENUE ===')
print(city_ranking)

# ==============================================================================
# TASK 12: Calculate monthly revenue and identify the best month
# ==============================================================================
monthly_revenue = df.groupby(['Month', 'Month_Name'])['Revenue'].sum()
best_month = monthly_revenue.idxmax()[1]
best_month_rev = monthly_revenue.max()
print('\n=== TASK 12: MONTHLY REVENUE ===')
print(monthly_revenue)
print(
    f'\nBest Performing Month: {best_month} (Revenue: ₹{best_month_rev:,.2f})'
)

# ==============================================================================
# TASK 13: Find the top 10 customers by total spending
# ==============================================================================
top_10_customers = (
    df.groupby('Customer_ID')['Revenue']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
print('\n=== TASK 13: TOP 10 CUSTOMERS BY SPENDING ===')
print(top_10_customers)

# ==============================================================================
# TASK 14: Compare payment methods by number of orders and revenue
# ==============================================================================
payment_comparison = df.groupby('Payment_Method').agg(Number_of_Orders=('Order_ID', 'size'), Total_Revenue=('Revenue', 'sum')
)
print('\n=== TASK 14: PAYMENT METHOD COMPARISON ===')
print(payment_comparison)

#==============================================================================
#TASK 15: Find unusually large order values using the IQR method
#==============================================================================

q1 = np.percentile(df['Revenue'], 25)
q3 = np.percentile(df['Revenue'], 75)
iqr = q3 - q1
upper_threshold = q3 + 1.5 * iqr
large_orders = df[df['Revenue'] > upper_threshold]
print('\n=== TASK 15: UNUSUALLY LARGE ORDERS (IQR METHOD) ===')
print(f'Q1: ₹{q1:,.2f} | Q3: ₹{q3:,.2f} | IQR: ₹{iqr:,.2f}')
print(f'Upper Outlier Threshold: ₹{upper_threshold:,.2f}')
print(large_orders[['Order_ID', 'Product', 'Quantity', 'Revenue']])

#==============================================================================
#TASK 16: Create a product-category summary using groupby
#==============================================================================

category_summary = df.groupby('Category').agg(
Orders=('Order_ID', 'size'),
Total_Quantity=('Quantity', 'sum'),
Total_Revenue=('Revenue', 'sum'),
)
print('\n=== TASK 16: PRODUCT CATEGORY SUMMARY ===')
print(category_summary)

#==============================================================================
#TASK 17: Use Numpy to calculate mean, median, and standard deviation of order revenue
#==============================================================================

revenue_array = np.array(df['Revenue'])
print('\n=== TASK 17: NUMPY REVENUE STATISTICS ===')
print(f'Mean Revenue:               ₹{np.mean(revenue_array):,.2f}')
print(f'Median Revenue:             ₹{np.median(revenue_array):,.2f}')
print(f'Standard Deviation Revenue: ₹{np.std(revenue_array):,.2f}')

#==============================================================================
#TASK 18: Write five business recommendations supported by your results
#==============================================================================

print('\n=== TASK 18: FIVE EVIDENCE-BASED BUSINESS RECOMMENDATIONS ===')
recommendations = [
f'1. Focus Stock & Ads on Top Products: Gaming Chairs (₹90,000) and Laptops (₹60,000) lead sales; prioritize inventory and targeted ads for top 10 items.',
f'2. Expand Regional Marketing in Top Cities: Delhi (₹119,000) and Mumbai (₹92,500) generate the highest revenue; run loyalty campaigns there while offering promos in Pune.',
f'3. Capitalize on Seasonal Peaks: {best_month} is the strongest month (₹{best_month_rev:,.2f}); increase ad spend and stock levels prior to Q4.',
f'4. Launch a VIP Program for High-Value & Bulk Buyers: Outliers above ₹{upper_threshold:,.2f} (such as bulk chair orders) significantly boost revenue; offer corporate/bulk discounts.',
f'5. Optimize Checkout for Credit Card & UPI: Credit Cards generate the highest revenue (₹205,000) while UPI leads in order frequency (6 orders); streamline both at checkout.',
]
for rec in recommendations:
    print(rec)
```
