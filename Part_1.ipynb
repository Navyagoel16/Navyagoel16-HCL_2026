```python
#Task 4.
# Store the six orders as a list of tuples: (item_name, quantity, unit_price)
orders = [
    ("Wireless Mouse", 2, 799),
    ("Basmati Rice 5kg", 8, 650),
    ("Denim Jeans", 7, 1599),
    ("Bluetooth Speaker", 5, 2499),
    ("Notebook Pack", 10, 120),
    ("USB-C Cable", 0, 299)  # invalid - must be skipped
]

grand_total = 0.0

# Loop through orders using enumerate() starting index at 1
for index, (item, qty, price) in enumerate(orders, start=1):
    # Skip invalid orders where quantity or price is 0 or negative
    if qty <= 0 or price <= 0:
        continue

    # Calculate subtotal
    subtotal = qty * price

    # Determine discount slab based on subtotal
    if subtotal >= 10000:
        discount_rate = 0.10
    elif subtotal >= 5000:
        discount_rate = 0.05
    else:
        discount_rate = 0.0

    discount = subtotal * discount_rate
    taxable_amount = subtotal - discount

    # Apply 18% GST on the discounted amount
    gst = taxable_amount * 0.18
    final_amount = taxable_amount + gst

    # Add to grand total
    grand_total += final_amount

    # Print formatted bill line
    print(f"Order {index} - {item}: Subtotal = Rs. {subtotal:.2f}, Discount = Rs. {discount:.2f}, GST = Rs. {gst:.2f}, Final = Rs. {final_amount:.2f}")

# Print the grand total of all valid orders
print(f"\nGrand Total: Rs. {grand_total:.2f}")
```

```python
#TASK 6.
# Sample catalogue list of products
catalogue = ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"]

# Ask the user to enter a product name
search_item = input("Enter the product name to search: ")

# Search the catalogue using a for-else loop
for index in range(len(catalogue)):
    # .lower() makes the search case-insensitive
    if catalogue[index].lower() == search_item.lower():
        # Adding 1 to index so position starts at 1 instead of 0
        print(f"Product found at position {index + 1}.")
        break
else:
    # Runs only if the loop finishes without hitting 'break'
    print("Not stocked.")
```
