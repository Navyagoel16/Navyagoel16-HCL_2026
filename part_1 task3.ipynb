```python
#Task 2: Slab discount using if, elif, else

# valid or not
quantity= 5
unit_price= 5000

if quantity > 0 and unit_price > 0:
    print("valid")


    # slab discount
    subtotal= quantity * unit_price

    if subtotal >= 10000:
        discount= subtotal * 0.1

    elif subtotal >= 5000:
        discount= subtotal* 0.05

    else:
        discount= 0

    final_total= subtotal - discount
    print("Subtotal:", subtotal)
    print("Discount:", discount)
    print("Final_toal:", final_total)

else:
    print("not valid")
```
