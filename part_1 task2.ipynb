```python
import math

# 1. Compute subtotal, 18% GST, and final amount
notebooks = 100
price_per_notebook = 50  # Sample unit price in INR

subtotal = notebooks * price_per_notebook
gst = subtotal * 0.18
final_amount = subtotal + gst

print("Subtotal:", subtotal)
print("18% GST:", gst)
print("Final Amount:", final_amount)

print("-------")

# 2. Express 100 notebooks as full cartons of 12 plus loose units
carton_size = 12
full_cartons = notebooks // carton_size
loose_units = notebooks % carton_size

print("Full cartons:", full_cartons)
print("Loose units:", loose_units)

print("-------")

# 3. Floating-point comparison and fix
print("0.1 + 0.2 == 0.3 is:", 0.1 + 0.2 == 0.3)
print("Fixed using round():", round(0.1 + 0.2, 2) == 0.3)
print("Fixed using math.isclose():", math.isclose(0.1 + 0.2, 0.3))
```
